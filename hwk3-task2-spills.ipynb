````python
# FINAL CODE

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import geopandas as gpd

# Load spills data
fp = os.path.join('data','Spill_Incidents_20261102.csv')
spills = pd.read_csv(fp)

# Load county lines data
fp2 = os.path.join('data','tl_2026_us_county','tl_2026_us_county.shp')
counties = gpd.read_file(fp2)

# changing column names to lower case then replacing spaces with underscore
spills.columns = spills.columns.str.lower().str.replace(' ','_')

# Update spills dates to datetime objects
spills[['spill_date','received_date','close_date']] = spills[['spill_date','received_date','close_date']].astype("datetime64[ns]")
# Remove commas from quantity and recovered to float
spills['quantity'] = spills['quantity'].replace(',','', regex = True)
spills['recovered'] = spills['recovered'].replace(',','', regex = True)
spills[['quantity','recovered']] = spills[['quantity','recovered']].astype('float64')

# Filter for petroleum spills in material_family, then subsect for date
petr_23 = spills[(spills['material_family'] == 'Petroleum') & spills['spill_date'].between('2023-01-01','2023-12-31')]

# Create new dataframe, grouping by county, counting the number of unique values
spills_per_county = petr_23.groupby('county').nunique('spill_number')['spill_number'].reset_index()

# Rename spills per county columns
spills_per_county = spills_per_county.rename(columns={'spill_number':'n_spills'})

# Counties column name to lower case
counties.columns = counties.columns.str.lower()

# Select NY counties, NY state number is 36
ny_counties = counties[counties['statefp'] == '36']

# find the index of St Lawrence
spills_per_county['county'].tolist().index('St Lawrence')
# Use loc to rename the cell
spills_per_county.loc[52,'county'] = 'St. Lawrence'
# Check that name was changed
assert spills_per_county.loc[52,'county'] == 'St. Lawrence'

# Inner merge spills and NY counties on 'county'
ny_spills = pd.merge(spills_per_county, ny_counties, left_on='county', right_on='name', how='inner')

# Subset data about Niagara and Onondaga counties
reservation_counties = ny_counties[ny_counties['name'].isin(['Niagara','Onondaga'])].reset_index()
# Subset to spills in the respective Reservations
reservation_spills = spills_per_county[spills_per_county['county'].isin(['Tuscarora Indian Reservation','Onondaga Indian Reservation'])].reset_index()
# Merge both data subsects, drop index column to match shape of ny_spills
reserv_join = pd.concat([reservation_spills, reservation_counties], axis = 1).drop('index',axis =1)

# Join the subsect to the main dataframe
ny_spills_updated = pd.concat([ny_spills, reserv_join])
# Adding two more rows should mean that the value of sum of TRUE is 2
assert sum(ny_spills_updated['county'].isin(['Tuscarora Indian Reservation','Onondaga Indian Reservation'])) == 2

# Turn dataframe into geodataframe
gdf = gpd.GeoDataFrame(ny_spills_updated, geometry="geometry")

# Initialize figure
plt.figure(figsize=(8,6))

# Plot the map + create labels
gdf.plot(column = 'n_spills',
         legend = True,
         legend_kwds={'label':'Number of spills'})

plt.title('Petroleum Spill Count Per County in New York in 2023')
plt.axis('off')
plt.annotate('Data Source: New York State Department of Environmental Conservation\n*Tuscarora and Onondaga Indian Reservations included.', 
            xy = (0, -0.1),xycoords='axes fraction',fontsize=9)

plt.tight_layout()
````
